### Required Assignment 5.1: Will the Customer Accept the Coupon?

**Context**

Imagine driving through town and a coupon is delivered to your cell phone for a restaurant near where you are driving. Would you accept that coupon and take a short detour to the restaurant? Would you accept the coupon but use it on a subsequent trip? Would you ignore the coupon entirely? What if the coupon was for a bar instead of a restaurant? What about a coffee house? Would you accept a bar coupon with a minor passenger in the car? What about if it was just you and your partner in the car? Would weather impact the rate of acceptance? What about the time of day?

Obviously, proximity to the business is a factor on whether the coupon is delivered to the driver or not, but what are the factors that determine whether a driver accepts the coupon once it is delivered to them? How would you determine whether a driver is likely to accept a coupon?

**Overview**

The goal of this project is to use what you know about visualizations and probability distributions to distinguish between customers who accepted a driving coupon versus those that did not.

**Data**

This data comes to us from the UCI Machine Learning repository and was collected via a survey on Amazon Mechanical Turk. The survey describes different driving scenarios including the destination, current time, weather, passenger, etc., and then ask the person whether he will accept the coupon if he is the driver. Answers that the user will drive there ‘right away’ or ‘later before the coupon expires’ are labeled as ‘Y = 1’ and answers ‘no, I do not want the coupon’ are labeled as ‘Y = 0’.  There are five different types of coupons -- less expensive restaurants (under \$20), coffee houses, carry out & take away, bar, and more expensive restaurants (\$20 - $50).

2. What proportion of bar coupons were accepted?

3. Compare the acceptance rate between those who went to a bar 3 or fewer times a month to those who went more.


```python
# get the unique value from bar_coupons
# print([bar_coupons['Bar'].unique()])
# ['never', 'less1', '1~3', 'gt8', '4~8']

# to extract 3 or fewer times a month
fewer_visit_to_bar = bar_coupons[bar_coupons['Bar'].isin(['never', 'less1', '1~3'])]['Y'].mean()


# more than 3 times a month
more_visit_to_bar = bar_coupons[bar_coupons['Bar'].isin(['4~8', 'gt8'])]['Y'].mean()

print(f"Acceptance rate (3 or fewer visits): {fewer_visit_to_bar:.2%}")
print(f"Acceptance rate (more than 3 visits): {more_visit_to_bar:.2%}")

```

4. Compare the acceptance rate between drivers who go to a bar more than once a month and are over the age of 25 to the all others.  Is there a difference?


```python
# Age is string column and converting to numeric
age_map = {
    'below21': 20,
    '21': 21,
    '26': 26,
    '31': 31,
    '36': 36,
    '41': 41,
    '46': 46,
    '50plus': 50
}
# create a new numberic column
bar_coupons['age_numeric'] = bar_coupons['age'].map(age_map)

over_25_bar_visitor = bar_coupons[
    (bar_coupons['age_numeric'] > 25) &
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8']))
]['Y'].mean()
# print(over_25_bar_visitor)
others = bar_coupons[
    ~((bar_coupons['age_numeric'] > 25) &
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])))
]['Y'].mean()

print(f"Sample population over 25yrs and go to bar more than once a month is {over_25_bar_visitor: .2%}")
print(f"all others are {others: .2%}")

# print(bar_coupons['age'])
```

5. Use the same process to compare the acceptance rate between drivers who go to bars more than once a month and had passengers that were not a kid and had occupations other than farming, fishing, or forestry.

6. Compare the acceptance rates between those drivers who:

- go to bars more than once a month, had passengers that were not a kid, and were not widowed *OR*
- go to bars more than once a month and are under the age of 30 *OR*
- go to cheap restaurants more than 4 times a month and income is less than 50K.



```python
maritalStatus = bar_coupons['maritalStatus'].unique()
income = bar_coupons['income'].unique()
RestaurantLessThan20 = bar_coupons['RestaurantLessThan20'].unique()
print(maritalStatus, income, RestaurantLessThan20)
```

```python
# condition1 - go to bars more than once a month, had passengers that were not a kid, and were not widowed *OR*
# condition2 - go to bars more than once a month and are under the age of 30 *OR*
# condition3 - go to cheap restaurants more than 4 times a month and income is less than 50K.

condition1 = (
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])) &
    (bar_coupons['passanger'] != 'Kid(s)') &
    (bar_coupons['maritalStatus'] != 'Widowed')
)

condition2 = (
    (bar_coupons['Bar'].isin(['1~3', '4~8', 'gt8'])) &
    (bar_coupons['age_numeric'] < 30)
)

condition3 = (
    (bar_coupons['RestaurantLessThan20'].isin(['4~8', 'gt8'])) &
    (bar_coupons['income'].isin(['Less than $12500', '$12500 - $24999', '$25000 - $37499', '$37500 - $49999']))
)

```

```python
groups = ['Condition1: No kid, not widowed', 'Condition2: Under 30', 'Condition3: Cheap restaurant & low income', 'All Others']
rates = [
    bar_coupons[condition1]['Y'].mean(),
    bar_coupons[condition2]['Y'].mean(),
    bar_coupons[condition3]['Y'].mean(),
    bar_coupons[~(condition1 | condition2 | condition3)]['Y'].mean()
]

sns.barplot(x=groups, y=rates, palette='Set2')
plt.title('Acceptance Rate by Driver Group')
plt.ylabel('Acceptance Rate')
plt.xticks(rotation=15, ha='right')
plt.tight_layout()
plt.savefig('images/bar_acceptance_by_frequency.png', bbox_inches='tight')
plt.show()
```

```python
# coffee house is at top and lets explore the statistics for coffee guys
coffee_coupons = data_cleaned[data_cleaned['coupon'] == 'Coffee House']
print(coffee_coupons.shape)
```

```python
coffee_coupon_acceptance_rate = coffee_coupons['Y'].mean()
print(f"Coffee House coupon acceptance rate is {coffee_coupon_acceptance_rate: .2%}")
```

```python
# like bar, lets use how frequent the drivers visit the coffee house in a given month
frequency_order = coffee_coupons['CoffeeHouse'].unique()
print(frequency_order)
frequency_order_sorted = ['never', 'less1', '1~3', '4~8', 'gt8']
```

```python
acceptance_by_freq = coffee_coupons.groupby('CoffeeHouse')['Y'].mean().reindex(frequency_order_sorted)
sns.barplot(x=acceptance_by_freq.index, y=acceptance_by_freq.values, palette='Set2')
plt.title('Coffee House Coupon Acceptance by Visit Frequency')
plt.xlabel('Monthly Visit Frequency')
plt.ylabel('Acceptance Rate')
plt.tight_layout()
plt.show()
```

The conclusion for above chart is very similar to bar goers; that is, frequent coffee house visitors are more likely to accept a coffee house coupon

```python
# lets use time of day
time_order = ['7AM', '10AM', '2PM', '6PM', '10PM'] # without this 7am comes in the end of the bar plot
acceptance_by_time = coffee_coupons.groupby('time')['Y'].mean().reindex(time_order)
sns.barplot(x=acceptance_by_time.index, y=acceptance_by_time.values, palette='Set2')
plt.title('Coffee House Coupon Acceptance by Time of Day')
plt.xlabel('Time of Day')
plt.ylabel('Acceptance Rate')
plt.savefig('images/coffee_acceptance_by_frequency.png', bbox_inches='tight')
plt.tight_layout()
plt.show()
```
